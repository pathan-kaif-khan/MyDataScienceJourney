## Mini Project: The Automated Receipt Cleaner

```python
# The Scenario:
# Imagine you work for an accounting firm. Clients send in text messages with their expenses,
# but the formatting is a complete mess. Your script needs to read this messy text, 
# extract the data dynamically, do some math, and print out a neat, professional receipt.

# Here is your messy input data:
# raw_receipt = "   STORE: TechMart || item: laptoP || price: $899.99 || tax: 8%   "
#Final output
#====================
# Store: Techmart
# Item: Laptop
# --------------------
# Base Price: $899.99
# Total (w/ Tax): $971.99
# ====================
```

```python
raw_receipt = "   STORE: TechMart || item: laptoP || price: $899.99 || tax: 8%   "
print("="*20)

store_start =raw_receipt.find("STORE") 
store_end = raw_receipt.find("|",store_start)
store=raw_receipt[store_start:store_end].strip().title()
print(store)

item_start =raw_receipt.find("item") 
item_end = raw_receipt.find("|",item_start)
item=raw_receipt[item_start:item_end].strip().title()
print(item)

print("-"*20)

price_start =raw_receipt.find("price") 
price_end = raw_receipt.find("|",price_start)
price = raw_receipt[price_start:price_end].strip().title()
print(f"Base {price}")

float_price_start =raw_receipt.find("$")+1 
float_price_end = raw_receipt.find("|",float_price_start)
float_price = raw_receipt[float_price_start:float_price_end].strip().title()
float_price = float(float_price)

tax_start = raw_receipt.find(":",float_price_end)+1
tax_end = raw_receipt.find("%")
tax = raw_receipt[tax_start:tax_end].strip()
tax =float(tax)

total = float_price+float_price*(tax/100)

print(f"Total (w/ Tax): ${total:.2f}")

print("="*20)
```

```python
# professional code 
raw_receipt = "   STORE: TechMart || item: laptoP || price: $899.99 || tax: 8%   "

# Clean the receipt
receipt = raw_receipt.strip()

# Extract values
store = receipt.split("||")[0].split(":", 1)[1].strip().title()
item = receipt.split("||")[1].split(":", 1)[1].strip().title()
price = float(receipt.split("||")[2].split("$")[1].strip())
tax = float(receipt.split("||")[3].split(":")[1].replace("%", "").strip())

# Calculate total
total = price * (1 + tax / 100)

# Display result
print("=" * 20)
print(f"Store: {store}")
print(f"Item: {item}")
print("-" * 20)
print(f"Base Price: ${price:.2f}")
print(f"Total (w/ Tax): ${total:.2f}")
```

Absolutely. That one line looks complicated because you are chaining several string methods together. Let's break it down from left to right.

Your code:

store = receipt.split("||")[0].split(":", 1)[1].strip().title()

1. Start with receipt

After:

receipt = raw_receipt.strip()


you have:

STORE: TechMart || item: laptoP || price: $899.99 || tax: 8%

2. .split("||")
receipt.split("||")


This separates the receipt wherever it sees ||.

You get a list:

[
    "STORE: TechMart ",
    " item: laptoP ",
    " price: $899.99 ",
    " tax: 8%"
]

3. [0]
receipt.split("||")[0]


We only want the first piece, which contains the store:

STORE: TechMart

4. .split(":", 1)

Now:

"STORE: TechMart".split(":", 1)


The 1 means split only once.

Result:

["STORE", " TechMart"]


We don't need "STORE"; we need " TechMart".

5. [1]
"STORE: TechMart".split(":", 1)[1]


gives:

 TechMart


Notice the space before TechMart.

6. .strip()
" TechMart".strip()


removes that extra space:

TechMart

7. .title()
"TechMart".title()


gives:

Techmart

So the whole chain is basically:
receipt
   ↓
split by ||
   ↓
take [0]
   ↓
split by :
   ↓
take [1]
   ↓
strip spaces
   ↓
title case
   ↓
"Techmart"


So:

store = receipt.split("||")[0].split(":", 1)[1].strip().title()


produces:

print(store)

Techmart


Tip: As a beginner, don't feel like you need to write chains this long immediately. You can write the same logic step-by-step:

parts = receipt.split("||")
store_part = parts[0]

store_parts = store_part.split(":", 1)
store_name = store_parts[1]

store = store_name.strip().title()


This is often easier to debug and understand while you're learning.
