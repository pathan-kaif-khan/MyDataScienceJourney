```python

user_email = "   DATA.SCIENTIST@IBM.COM   "
clean_user_email = user_email.strip().lower()
at_index = clean_user_email.find("@")
domain = clean_user_email[at_index +1:]
print(domain)
```

```python
# Extract just the number portion ("1,450.99") from the string. You can use slicing, or the .find() trick we used earlier.

# If you try to convert a string with a comma into a number, Python will throw an error. Use a string method to replace the comma "," with an empty string "".

# Convert your clean string into a float (since it has decimals, int() won't work!).

# Apply a 10% discount to this price (multiply it by 0.9) and print the final discounted price.
```

```python
raw_price = "Total cost: $1,450.99 USD"
ind_start = raw_price.find("$") +1
ind_end = raw_price.find("U")

num_price = raw_price[ind_start:ind_end]
num_price #'1,450.99 '
num_price = num_price.replace(",","")
num_price #'1450.99 '
float_price = float(num_price)
discount_price = (float_price * 0.9)
print(discount_price)

"""
much more clean code for above 
raw_price = "Total cost: $1,450.99 USD"

start = raw_price.find("$") + 1
end = raw_price.find(" ", start)

price = float(raw_price[start:end].replace(",", ""))
discounted_price = price * 0.9

print(discounted_price)
"""
```
