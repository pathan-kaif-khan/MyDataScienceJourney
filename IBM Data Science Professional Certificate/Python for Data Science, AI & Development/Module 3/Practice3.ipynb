## Q1

```python
def greatest(a,b,c):
    if(a>b and a>c):
        return a
    elif(b>c and b>a):
        return b
    elif(c>a and c>b):
        return c
    
a = 25
b = 35
c = 8
print(greatest(a,b,c))
```

```python
#celsius to Fahrenheit

def f_to_c(f):
    return 5*(f-32)/9

f = int(input("enter temprature in F"))
c = f_to_c(f)
print(f"{round(c,2)}degree C")
```
