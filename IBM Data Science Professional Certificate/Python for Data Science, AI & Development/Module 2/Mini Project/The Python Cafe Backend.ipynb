## MINI PROJECT [The Python Cafe Backend]

```python
menu = [("espresso",2.99),("cappuccino",2.50),("latte",1.99)]
inventory = {"coffee beans" : 10, "milk" : 20, "syrup" : 15}
order_queue = []
daily_customers = set()

daily_customers.add("Alice")
order_queue.append("espresso")
inventory["coffee beans"] -= 1

daily_customers.add("Alice")
daily_customers.add("Bob")
order_queue.append("latte")
order_queue.append("cappuccino")

print(daily_customers)
print(order_queue)

total_revenue = 0
menu_dict = dict(menu)

for order in order_queue:
        total_revenue += menu_dict[order]
print(total_revenue)
```
