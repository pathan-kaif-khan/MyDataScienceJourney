## Q3

```python
prices = {"apple": 1.50, "banana": 0.50, "orange": 1.20}
cart = {"apple": 3, "banana": 5, "orange": 1}
total_cost = 0
for item in cart.items():
    x = item[1] #quantity
    price = prices[item[0]] #price
    total_cost = total_cost + (x * price)
total_cost 

```

```python
prices = {"apple": 1.50, "banana": 0.50, "orange": 1.20}
cart = {"apple": 3, "banana": 5, "orange": 1}

total_cost = 0

for item , quantity in cart.items():
    price = prices[item]
    total_cost += price * quantity

print(total_cost)
```

```python
posts = [
    {"title": "Learning Python", "likes": 15, "tags": ["coding", "python"]},
    {"title": "Had a great lunch", "likes": 5, "tags": ["food", "life"]},
    {"title": "Dictionaries are fun", "likes": 25, "tags": ["coding", "python", "learning"]}
]
total_coding_likes = 0 
for item in posts:
    if "coding" in item["tags"]:
        total_coding_likes += item["likes"]
print(total_coding_likes)
```
